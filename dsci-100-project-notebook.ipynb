```r
# histogram showing the count of conference win percentage
plot1 <- tidy_data |>
    ggplot(aes(x = conf_wins_pct)) +
    geom_histogram(bins = 20, fill = "steelblue", color = "white") +
    labs(title = "Distribution of Conference Win Percentage", 
         x = "Conference Win Percentage", y = "Count") +
    theme(text = element_text(size = 18))
plot1
```

```r
# bar plot showing proportion of different conferences
plot2 <- tidy_data |>
    ggplot(aes(x = conference)) +
    geom_bar(fill = "pink") +
    labs(title = "Proportion of Conferences", x = "Conference", y = "Count") +
    theme(axis.text.x = element_text(angle = 90, hjust = 1, vjust = .5, size = 8),
         text = element_text(size = 18))
plot2
```

According to visualization 1, the plot shows a weak negative relationship between conference wins and conference rank. As the number of wins increase, the conference rank decreases(meaning a better rank). According to the colors of the plot, conferences with higher rank are more likely to have an auto bid, while when conference ranking is around 5, the chance of receiving an at-large bid is higher.

```r
# visualization of conference rank vs conference wins using a scatterplot
visualization1 <- tidy_data |>
    ggplot(aes(x = conf_wins, y = conf_rank, color = bid)) + # use color to separate at-large and auto bid
    geom_point(alpha = .2) +
    labs(title ="Conference Rank vs. Conference Wins", 
         x = "Conference Wins", y = "Conference Rank", color = "Bid") +
    theme(text = element_text(size = 18))
visualization1
```

Overall, conferences with higher wins percentage and higher seed number are more likely to recieve auto bids.

```r
# visualization of conference wins percentage vs seed using a scatterplot
visualization2 <- tidy_data |>
    ggplot(aes(x = seed, y = conf_wins_pct, color = bid)) +
    geom_point(alpha = .2) +
    labs(title = "Conference Wins Percentage vs. Seed", 
         x = "Seed", y = "Conference Wins Percentage", color = "Bid") +
    theme(text = element_text(size = 18))
visualization2
```
